# Pandas Sereies
* A Pandas Series is like a column in a table.it is a 1-D array holding data of any type.

# Series from list

```python
# string = object in pandas
country = ["india","pakisthan","USA","Nepal","Sreilanka"]
pd.Series(country)
```

```python
# intergers
runs = [13,23,4,5,45,23,22,45]
pd.Series(runs)
```

```python
# custom index
marks = [67,57,89,100]
subjects = ['maths','english','science','hindi']
pd.Series(marks,index=subjects)
```

```python
# setting a name
marks = pd.Series(marks,index=subjects,name="poonam che marks")
marks
```

# Series from dict

```python
marks = {
    "maths":67,
    "english":57,
    "science":89,
    "hindi":100
}
m = pd.Series(marks,name = "poonam che marks")
m
```
